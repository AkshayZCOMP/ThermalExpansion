```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# USER SETTINGS
# ============================================================

# Put the FULL path to your CSV here
FILE_PATH = r"C:\Users\prz\Documents\HeatXfireDIC\HeatXferDIC\AbaqusSweep\ABaqusData_with_computational_results.csv"

# Curvature columns to plot
CURVATURE_COLUMNS = ["K_11", "K_22", "K_12"]

# If True, plots |curvature| instead of signed curvature
USE_ABSOLUTE_CURVATURE = False

# Size of points on distribution plot
POINT_SIZE = 30

# Transparency of individual points
POINT_ALPHA = 0.55


# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(FILE_PATH)

# Make sure required columns exist
required_columns = ["NumberOfPlies"] + CURVATURE_COLUMNS

missing = [col for col in required_columns if col not in df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

# Keep only rows with valid ply counts
df = df.dropna(subset=["NumberOfPlies"]).copy()

# Sort by ply count
df = df.sort_values("NumberOfPlies")

# Optionally use absolute curvature
if USE_ABSOLUTE_CURVATURE:
    for col in CURVATURE_COLUMNS:
        df[col] = df[col].abs()


# ============================================================
# PRINT RANGE TABLE
# ============================================================

range_table = df.groupby("NumberOfPlies")[CURVATURE_COLUMNS].agg(
    ["min", "max", "mean", "std", "count"]
)

display(range_table)


# ============================================================
# PLOT 1:
# MIN/MAX CURVATURE RANGE FOR EACH PLY COUNT
# ============================================================

fig, axes = plt.subplots(
    len(CURVATURE_COLUMNS),
    1,
    figsize=(10, 10),
    sharex=True
)

for ax, curvature in zip(axes, CURVATURE_COLUMNS):

    grouped = df.groupby("NumberOfPlies")[curvature]

    ply_counts = np.array(sorted(df["NumberOfPlies"].unique()))

    minimum = grouped.min().reindex(ply_counts).values
    maximum = grouped.max().reindex(ply_counts).values
    mean = grouped.mean().reindex(ply_counts).values

    # Vertical line = total min/max range
    ax.vlines(
        ply_counts,
        minimum,
        maximum,
        linewidth=4,
        alpha=0.55
    )

    # Min and max markers
    ax.scatter(
        ply_counts,
        minimum,
        marker="_",
        s=150,
        label="Minimum"
    )

    ax.scatter(
        ply_counts,
        maximum,
        marker="_",
        s=150,
        label="Maximum"
    )

    # Mean marker
    ax.scatter(
        ply_counts,
        mean,
        marker="o",
        s=45,
        label="Mean",
        zorder=3
    )

    ax.axhline(0, linewidth=0.8, alpha=0.5)

    ax.set_ylabel(f"{curvature}\n[1/m]")
    ax.set_title(f"{curvature} Range by Ply Count")
    ax.grid(alpha=0.25)

axes[-1].set_xlabel("Number of Plies")
axes[-1].set_xticks(sorted(df["NumberOfPlies"].unique()))

axes[0].legend()

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2:
# ALL CURVATURE RESULTS BY PLY COUNT
# ============================================================

fig, axes = plt.subplots(
    len(CURVATURE_COLUMNS),
    1,
    figsize=(10, 10),
    sharex=True
)

rng = np.random.default_rng(42)

for ax, curvature in zip(axes, CURVATURE_COLUMNS):

    for ply_count in sorted(df["NumberOfPlies"].unique()):

        values = df.loc[
            df["NumberOfPlies"] == ply_count,
            curvature
        ].dropna().values

        # Small horizontal jitter so overlapping points can be seen
        jitter = rng.uniform(-0.12, 0.12, len(values))

        ax.scatter(
            ply_count + jitter,
            values,
            s=POINT_SIZE,
            alpha=POINT_ALPHA
        )

    ax.axhline(0, linewidth=0.8, alpha=0.5)

    ax.set_ylabel(f"{curvature}\n[1/m]")
    ax.set_title(f"{curvature} Distribution by Ply Count")
    ax.grid(alpha=0.25)

axes[-1].set_xlabel("Number of Plies")
axes[-1].set_xticks(sorted(df["NumberOfPlies"].unique()))

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 3:
# COMBINED CURVATURE MAGNITUDE
#
# Useful if you care about whether the TOTAL curvature response
# distinguishes ply counts rather than an individual component.
# ============================================================

df["K_magnitude"] = np.sqrt(
    df["K_11"]**2 +
    df["K_22"]**2 +
    df["K_12"]**2
)

fig, ax = plt.subplots(figsize=(10, 6))

for ply_count in sorted(df["NumberOfPlies"].unique()):

    values = df.loc[
        df["NumberOfPlies"] == ply_count,
        "K_magnitude"
    ].dropna().values

    jitter = rng.uniform(-0.12, 0.12, len(values))

    ax.scatter(
        ply_count + jitter,
        values,
        s=POINT_SIZE,
        alpha=POINT_ALPHA
    )

ax.set_xlabel("Number of Plies")
ax.set_ylabel(r"Curvature Magnitude [1/m]")
ax.set_title("Total Curvature Magnitude by Ply Count")
ax.set_xticks(sorted(df["NumberOfPlies"].unique()))
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# USER SETTINGS
# ============================================================

FILE_PATH = r"C:\Users\prz\Documents\HeatXfireDIC\HeatXferDIC\AbaqusSweep\ABaqusData_with_computational_results.csv"

# Change these ONLY if your CSV uses different column names
FEA_COLUMNS = {
    "K11": "K_11",
    "K22": "K_22",
    "K12": "K_12",
}

COMPUTATIONAL_COLUMNS = {
    "K11": "Comp_K_11",
    "K22": "Comp_K_22",
    "K12": "Comp_K_12",
}


# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(FILE_PATH)

print("Columns in CSV:")
print(df.columns.tolist())


# ============================================================
# CALCULATE ERRORS
# ============================================================

for component in ["K11", "K22", "K12"]:

    fea_col = FEA_COLUMNS[component]
    comp_col = COMPUTATIONAL_COLUMNS[component]

    # Signed error
    df[f"{component}_Error"] = (
        df[fea_col] - df[comp_col]
    )

    # Absolute error
    df[f"{component}_AbsError"] = (
        df[f"{component}_Error"].abs()
    )

    # Percent error relative to FEA
    # NaN used when FEA is essentially zero
    denominator = df[fea_col].abs()

    df[f"{component}_PercentError"] = np.where(
        denominator > 1e-12,
        100 * df[f"{component}_AbsError"] / denominator,
        np.nan
    )


# ============================================================
# PRINT ERROR STATISTICS
# ============================================================

print("\n================ ERROR SUMMARY ================\n")

for component in ["K11", "K22", "K12"]:

    abs_error = df[f"{component}_AbsError"]
    percent_error = df[f"{component}_PercentError"]

    print(component)
    print(f"Mean absolute error:   {abs_error.mean():.6g} 1/m")
    print(f"Maximum absolute error:{abs_error.max():.6g} 1/m")
    print(f"Mean percent error:    {percent_error.mean():.3f} %")
    print(f"Maximum percent error: {percent_error.max():.3f} %")
    print()


# ============================================================
# PLOT 1
# ABSOLUTE ERROR FOR EVERY MODEL
# ============================================================

fig, axes = plt.subplots(
    3,
    1,
    figsize=(11, 10),
    sharex=True
)

components = ["K11", "K22", "K12"]

for ax, component in zip(axes, components):

    ax.scatter(
        df.index,
        df[f"{component}_AbsError"],
        s=25,
        alpha=0.7
    )

    # Mean error line
    mean_error = df[f"{component}_AbsError"].mean()

    ax.axhline(
        mean_error,
        linestyle="--",
        linewidth=1.5,
        label=f"Mean = {mean_error:.3e}"
    )

    ax.set_ylabel("Absolute Error [1/m]")
    ax.set_title(f"{component} â€” FEA vs Computational Error")

    ax.grid(alpha=0.25)
    ax.legend()

axes[-1].set_xlabel("Laminate / Simulation Number")

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2
# PERCENT ERROR FOR EVERY MODEL
# ============================================================

fig, axes = plt.subplots(
    3,
    1,
    figsize=(11, 10),
    sharex=True
)

for ax, component in zip(axes, components):

    ax.scatter(
        df.index,
        df[f"{component}_PercentError"],
        s=25,
        alpha=0.7
    )

    mean_error = df[f"{component}_PercentError"].mean()

    ax.axhline(
        mean_error,
        linestyle="--",
        linewidth=1.5,
        label=f"Mean = {mean_error:.2f}%"
    )

    ax.set_ylabel("Percent Error [%]")
    ax.set_title(f"{component} â€” Percent Error")

    ax.grid(alpha=0.25)
    ax.legend()

axes[-1].set_xlabel("Laminate / Simulation Number")

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 3
# ERROR VS NUMBER OF PLIES
# ============================================================

if "NumberOfPlies" in df.columns:

    fig, axes = plt.subplots(
        3,
        1,
        figsize=(10, 10),
        sharex=True
    )

    rng = np.random.default_rng(42)

    for ax, component in zip(axes, components):

        for ply_count in sorted(df["NumberOfPlies"].unique()):

            values = df.loc[
                df["NumberOfPlies"] == ply_count,
                f"{component}_AbsError"
            ].dropna()

            # Jitter points slightly so overlapping results are visible
            jitter = rng.uniform(-0.12, 0.12, len(values))

            ax.scatter(
                ply_count + jitter,
                values,
                s=30,
                alpha=0.6
            )

        ax.set_ylabel("Absolute Error [1/m]")
        ax.set_title(f"{component} Error by Ply Count")
        ax.grid(alpha=0.25)

    axes[-1].set_xlabel("Number of Plies")
    axes[-1].set_xticks(
        sorted(df["NumberOfPlies"].unique())
    )

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# USER SETTINGS
# ============================================================

FILE_PATH = r"C:\Users\prz\Documents\HeatXfireDIC\HeatXferDIC\AbaqusSweep\ABaqusData_with_computational_results.csv"

# Change these if your CSV uses different column names
FEA_COLUMNS = {
    "K11": "K_11",
    "K22": "K_22",
    "K12": "K_12",
}

COMPUTATIONAL_COLUMNS = {
    "K11": "Comp_K_11",
    "K22": "Comp_K_22",
    "K12": "Comp_K_12",
}


# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(FILE_PATH)

print("Columns in CSV:")
print(df.columns.tolist())

components = ["K11", "K22", "K12"]


# ============================================================
# CALCULATE INDIVIDUAL CURVATURE ERRORS
# ============================================================

for component in components:

    fea_col = FEA_COLUMNS[component]
    comp_col = COMPUTATIONAL_COLUMNS[component]

    # Signed error
    df[f"{component}_Error"] = (
        df[fea_col] - df[comp_col]
    )

    # Absolute error
    df[f"{component}_AbsError"] = (
        df[f"{component}_Error"].abs()
    )

    # Percent error
    denominator = df[fea_col].abs()

    df[f"{component}_PercentError"] = np.where(
        denominator > 1e-12,
        100 * df[f"{component}_AbsError"] / denominator,
        np.nan
    )


# ============================================================
# COMBINED CURVATURE-VECTOR ERROR
# ============================================================

# Error vector:
#
# e = [K11_FEA - K11_Comp,
#      K22_FEA - K22_Comp,
#      K12_FEA - K12_Comp]
#
# Magnitude:
#
# ||e|| = sqrt(e11^2 + e22^2 + e12^2)

df["Curvature_Vector_Error"] = np.sqrt(
    df["K11_Error"]**2
    + df["K22_Error"]**2
    + df["K12_Error"]**2
)


# ============================================================
# FEA CURVATURE-VECTOR MAGNITUDE
# ============================================================

df["FEA_Curvature_Magnitude"] = np.sqrt(
    df[FEA_COLUMNS["K11"]]**2
    + df[FEA_COLUMNS["K22"]]**2
    + df[FEA_COLUMNS["K12"]]**2
)


# ============================================================
# COMPUTATIONAL CURVATURE-VECTOR MAGNITUDE
# ============================================================

df["Comp_Curvature_Magnitude"] = np.sqrt(
    df[COMPUTATIONAL_COLUMNS["K11"]]**2
    + df[COMPUTATIONAL_COLUMNS["K22"]]**2
    + df[COMPUTATIONAL_COLUMNS["K12"]]**2
)


# ============================================================
# NORMALIZED CURVATURE-VECTOR ERROR
# ============================================================
#
# This tells us how large the vector error is relative to the
# total magnitude of the FEA curvature vector.
#
# Relative error (%) =
#
#       ||K_FEA - K_Comp||
# 100 * -----------------
#           ||K_FEA||
#

df["Curvature_Vector_Percent_Error"] = np.where(
    df["FEA_Curvature_Magnitude"] > 1e-12,
    100
    * df["Curvature_Vector_Error"]
    / df["FEA_Curvature_Magnitude"],
    np.nan
)


# ============================================================
# PRINT INDIVIDUAL COMPONENT ERROR STATISTICS
# ============================================================

print("\n")
print("=" * 60)
print("INDIVIDUAL CURVATURE COMPONENT ERRORS")
print("=" * 60)

for component in components:

    abs_error = df[f"{component}_AbsError"]
    percent_error = df[f"{component}_PercentError"]

    print(f"\n{component}")
    print(f"Mean absolute error:     {abs_error.mean():.6e} 1/m")
    print(f"Median absolute error:   {abs_error.median():.6e} 1/m")
    print(f"Maximum absolute error:  {abs_error.max():.6e} 1/m")

    print(f"Mean percent error:      {percent_error.mean():.3f} %")
    print(f"Median percent error:    {percent_error.median():.3f} %")
    print(f"Maximum percent error:   {percent_error.max():.3f} %")


# ============================================================
# PRINT COMBINED VECTOR ERROR STATISTICS
# ============================================================

print("\n")
print("=" * 60)
print("COMBINED CURVATURE-VECTOR ERROR")
print("=" * 60)

print(
    f"Mean vector error:       "
    f"{df['Curvature_Vector_Error'].mean():.6e} 1/m"
)

print(
    f"Median vector error:     "
    f"{df['Curvature_Vector_Error'].median():.6e} 1/m"
)

print(
    f"Maximum vector error:    "
    f"{df['Curvature_Vector_Error'].max():.6e} 1/m"
)

print()

print(
    f"Mean normalized error:   "
    f"{df['Curvature_Vector_Percent_Error'].mean():.3f} %"
)

print(
    f"Median normalized error: "
    f"{df['Curvature_Vector_Percent_Error'].median():.3f} %"
)

print(
    f"Maximum normalized error:"
    f" {df['Curvature_Vector_Percent_Error'].max():.3f} %"
)


# ============================================================
# PLOT 1
# COMBINED VECTOR ERROR FOR EVERY LAMINATE
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

ax.scatter(
    df.index,
    df["Curvature_Vector_Error"],
    s=30,
    alpha=0.7
)

mean_error = df["Curvature_Vector_Error"].mean()
median_error = df["Curvature_Vector_Error"].median()

ax.axhline(
    mean_error,
    linestyle="--",
    linewidth=1.5,
    label=f"Mean = {mean_error:.3e} 1/m"
)

ax.axhline(
    median_error,
    linestyle=":",
    linewidth=1.5,
    label=f"Median = {median_error:.3e} 1/m"
)

ax.set_xlabel("Laminate / Simulation Number")
ax.set_ylabel("Curvature Vector Error [1/m]")

ax.set_title(
    "Combined FEA vs Computational Curvature Error"
)

ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2
# NORMALIZED VECTOR ERROR FOR EVERY LAMINATE
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

ax.scatter(
    df.index,
    df["Curvature_Vector_Percent_Error"],
    s=30,
    alpha=0.7
)

mean_percent = df["Curvature_Vector_Percent_Error"].mean()
median_percent = df["Curvature_Vector_Percent_Error"].median()

ax.axhline(
    mean_percent,
    linestyle="--",
    linewidth=1.5,
    label=f"Mean = {mean_percent:.2f}%"
)

ax.axhline(
    median_percent,
    linestyle=":",
    linewidth=1.5,
    label=f"Median = {median_percent:.2f}%"
)

ax.set_xlabel("Laminate / Simulation Number")
ax.set_ylabel("Normalized Curvature Error [%]")

ax.set_title(
    "Normalized FEA vs Computational Curvature-Vector Error"
)

ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 3
# COMBINED VECTOR ERROR BY PLY COUNT
# ============================================================

if "NumberOfPlies" in df.columns:

    fig, ax = plt.subplots(figsize=(10, 6))

    rng = np.random.default_rng(42)

    ply_counts = sorted(df["NumberOfPlies"].dropna().unique())

    for ply_count in ply_counts:

        values = df.loc[
            df["NumberOfPlies"] == ply_count,
            "Curvature_Vector_Error"
        ].dropna()

        jitter = rng.uniform(
            -0.12,
            0.12,
            len(values)
        )

        ax.scatter(
            ply_count + jitter,
            values,
            s=35,
            alpha=0.65
        )

    ax.set_xlabel("Number of Plies")
    ax.set_ylabel("Curvature Vector Error [1/m]")

    ax.set_title(
        "Combined Curvature-Vector Error by Ply Count"
    )

    ax.set_xticks(ply_counts)
    ax.grid(alpha=0.25)

    plt.tight_layout()
    plt.show()


# ============================================================
# PLOT 4
# NORMALIZED VECTOR ERROR BY PLY COUNT
# ============================================================

if "NumberOfPlies" in df.columns:

    fig, ax = plt.subplots(figsize=(10, 6))

    rng = np.random.default_rng(42)

    ply_counts = sorted(df["NumberOfPlies"].dropna().unique())

    for ply_count in ply_counts:

        values = df.loc[
            df["NumberOfPlies"] == ply_count,
            "Curvature_Vector_Percent_Error"
        ].dropna()

        jitter = rng.uniform(
            -0.12,
            0.12,
            len(values)
        )

        ax.scatter(
            ply_count + jitter,
            values,
            s=35,
            alpha=0.65
        )

    ax.set_xlabel("Number of Plies")
    ax.set_ylabel("Normalized Curvature Error [%]")

    ax.set_title(
        "Normalized Curvature-Vector Error by Ply Count"
    )

    ax.set_xticks(ply_counts)
    ax.grid(alpha=0.25)

    plt.tight_layout()
    plt.show()


# ============================================================
# PLOT 5
# FEA VS COMPUTATIONAL CURVATURE MAGNITUDE
# ============================================================

fig, ax = plt.subplots(figsize=(7, 7))

ax.scatter(
    df["FEA_Curvature_Magnitude"],
    df["Comp_Curvature_Magnitude"],
    s=35,
    alpha=0.7
)

# Perfect agreement line
all_values = np.concatenate([
    df["FEA_Curvature_Magnitude"].dropna().values,
    df["Comp_Curvature_Magnitude"].dropna().values
])

minimum = all_values.min()
maximum = all_values.max()

ax.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--",
    linewidth=1.5,
    label="Perfect agreement"
)

ax.set_xlabel("FEA Curvature Magnitude [1/m]")
ax.set_ylabel("Computational Curvature Magnitude [1/m]")

ax.set_title(
    "FEA vs Computational Curvature Magnitude"
)

ax.set_aspect("equal", adjustable="box")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STABLE ERROR METRICS: CURVATURE AND MIDPLANE STRAIN
# ============================================================
#
# Do not use per-case |error| / |FEA| as the headline metric near
# zero: an accurate small response still has a very small denominator.
# This cell reports several complementary measures instead:
#   1. absolute error (physical units),
#   2. symmetric percentage error (bounded at 200%), and
#   3. error normalized by a fixed data-set response scale.
# The last measure is the recommended relative metric for comparison.

RESPONSE_GROUPS = {
    "Curvature": {
        "unit": "1/m",
        "components": {
            "K11": ("K_11", "Comp_K_11"),
            "K22": ("K_22", "Comp_K_22"),
            "K12": ("K_12", "Comp_K_12"),
        },
    },
    "Strain": {
        "unit": "strain",
        "components": {
            "e011": ("e0_11", "Comp_e0_11"),
            "e022": ("e0_22", "Comp_e0_22"),
            "e012": ("e0_12", "Comp_e0_12"),
        },
    },
}

# A 95th-percentile magnitude is resistant to isolated extremes.  The
# 1% floor avoids treating numerical noise close to zero as a large error.
SCALE_PERCENTILE = 95
DENOMINATOR_FLOOR_FRACTION = 0.01

def robust_error_metrics(reference, prediction):
    """Return error series and a fixed, data-derived response scale."""
    absolute_error = (reference - prediction).abs()
    combined_magnitudes = np.concatenate([reference.abs().to_numpy(),
                                        prediction.abs().to_numpy()])
    scale = np.nanpercentile(combined_magnitudes, SCALE_PERCENTILE)
    scale = max(scale, np.finfo(float).eps)
    floor = max(DENOMINATOR_FLOOR_FRACTION * scale, np.finfo(float).eps)

    # Conventional percent error is diagnostic only; it is unstable near zero.
    conventional = 100 * absolute_error / reference.abs().where(reference.abs() > floor)

    # sMAPE is bounded, but remains best used alongside absolute error.
    symmetric_denominator = reference.abs() + prediction.abs()
    symmetric = (200 * absolute_error / symmetric_denominator.where(
        symmetric_denominator > floor
    ))

    # Recommended: a common scale for every case, including near-zero values.
    scale_normalized = 100 * absolute_error / scale
    floored_reference = 100 * absolute_error / np.maximum(reference.abs(), floor)

    return absolute_error, conventional, symmetric, scale_normalized, floored_reference, scale, floor

summary_rows = []

for group_name, group in RESPONSE_GROUPS.items():
    available = {
        name: columns for name, columns in group["components"].items()
        if set(columns).issubset(df.columns)
    }
    if not available:
        print(f"Skipping {group_name}: required columns were not found.")
        continue

    for component, (fea_col, comp_col) in available.items():
        (absolute, conventional, symmetric, scale_normalized, floored,
         scale, floor) = robust_error_metrics(df[fea_col], df[comp_col])
        prefix = f"{group_name}_{component}"
        df[f"{prefix}_AbsoluteError"] = absolute
        df[f"{prefix}_ConventionalPercentError"] = conventional
        df[f"{prefix}_SymmetricPercentError"] = symmetric
        df[f"{prefix}_ScaleNormalizedPercentError"] = scale_normalized
        df[f"{prefix}_FlooredPercentError"] = floored
        df[f"{prefix}_NearZero"] = df[fea_col].abs() < floor

        summary_rows.append({
            "response": group_name,
            "component": component,
            "unit": group["unit"],
            "scale (p95 |response|)": scale,
            "near-zero floor": floor,
            "near-zero cases": int((df[fea_col].abs() < floor).sum()),
            "MAE": absolute.mean(),
            "median sMAPE [%]": symmetric.median(),
            "median scale-normalized error [%]": scale_normalized.median(),
            "mean scale-normalized error [%]": scale_normalized.mean(),
        })

    fea_matrix = np.column_stack([df[columns[0]] for columns in available.values()])
    comp_matrix = np.column_stack([df[columns[1]] for columns in available.values()])
    vector_error = np.linalg.norm(fea_matrix - comp_matrix, axis=1)
    vector_scale = max(np.nanpercentile(np.linalg.norm(fea_matrix, axis=1),
                             SCALE_PERCENTILE), np.finfo(float).eps)
    fea_magnitude = np.linalg.norm(fea_matrix, axis=1)
    comp_magnitude = np.linalg.norm(comp_matrix, axis=1)
    vector_floor = max(DENOMINATOR_FLOOR_FRACTION * vector_scale,
                       np.finfo(float).eps)
    df[f"{group_name}_VectorAbsoluteError"] = vector_error
    df[f"{group_name}_VectorConventionalPercentError"] = (
        100 * vector_error / np.where(fea_magnitude > vector_floor,
                                      fea_magnitude, np.nan)
    )
    symmetric_vector_denominator = fea_magnitude + comp_magnitude
    df[f"{group_name}_VectorSymmetricPercentError"] = (
        200 * vector_error / np.where(
            symmetric_vector_denominator > vector_floor,
            symmetric_vector_denominator, np.nan
        )
    )
    df[f"{group_name}_VectorScaleNormalizedPercentError"] = (
        100 * vector_error / vector_scale
    )

error_summary = pd.DataFrame(summary_rows)
print("Recommended headline metric: scale-normalized error (not conventional percent error).\n")
display(error_summary.round(4))

# Compare the three percentage-style vector metrics directly.
PLOT_METRICS = [
    ("ConventionalPercentError", "Conventional relative error"),
    ("SymmetricPercentError", "Symmetric error (sMAPE)"),
    ("ScaleNormalizedPercentError", "Fixed-scale normalized error"),
]
fig, axes = plt.subplots(2, 3, figsize=(16, 8), sharex=True)
for row, group_name in enumerate(["Curvature", "Strain"]):
    for ax, (suffix, title) in zip(axes[row], PLOT_METRICS):
        vector_column = f"{group_name}_Vector{suffix}"
        if vector_column not in df:
            ax.set_visible(False)
            continue
        values = df[vector_column]
        ax.scatter(df.index, values, s=24, alpha=0.7)
        ax.axhline(values.median(), color="tab:red", linestyle="--",
                   label=f"Median = {values.median():.2f}%")
        ax.set_title(title)
        ax.set_ylabel(f"{group_name} vector error [%]")
        ax.grid(alpha=0.25)
        ax.legend()
for ax in axes[-1]:
    ax.set_xlabel("Laminate / Simulation Number")
fig.suptitle("Comparison of three curvature and strain error metrics")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# COMPONENT ERROR WITH NEAR-ZERO RESPONSES IGNORED
# ============================================================
#
# If the FEA strain/curvature component is essentially zero, this cell
# sets that component error to zero before plotting. That avoids letting
# tiny physical responses dominate the percent-error plots.

# Separate physical-zero cutoff for every response component.
COMPONENT_ZERO_THRESHOLDS = {
    "e0_11": 1e-8,   # strain
    "e0_22": 1e-8,   # strain
    "e0_12": 1e-10,  # strain
    "K_11": 1e-5,    # 1/m
    "K_22": 1e-5,    # 1/m
    "K_12": 1e-5,    # 1/m
}
COMPONENT_ERROR_METRIC = "ScaleNormalizedPercentError"

if "df" not in globals():
    raise NameError("Run the stable error metrics cell above this one first so df contains the error columns.")

near_zero_component_rows = []

for group_name, group in RESPONSE_GROUPS.items():
    for component_name, (fea_col, comp_col) in group["components"].items():
        error_col = f"{group_name}_{component_name}_{COMPONENT_ERROR_METRIC}"
        if error_col not in df.columns:
            print(f"Skipping missing column: {error_col}")
            continue
        if fea_col not in df.columns:
            print(f"Skipping missing FEA column: {fea_col}")
            continue

        threshold = COMPONENT_ZERO_THRESHOLDS[fea_col]
        near_zero_response = df[fea_col].abs() < threshold
        plot_values = df[error_col].copy()
        plot_values = plot_values.mask(near_zero_response, 0.0)

        near_zero_component_rows.append(
            pd.DataFrame(
                {
                    "Group": group_name,
                    "Component": component_name,
                    "CaseIndex": df.index,
                    "FEAValue": df[fea_col],
                    "ComputationalValue": df[comp_col],
                    "ZeroThreshold": threshold,
                    "NearZeroResponse": near_zero_response,
                    "OriginalErrorPercent": df[error_col],
                    "NearZeroAdjustedErrorPercent": plot_values,
                }
            )
        )

if not near_zero_component_rows:
    raise ValueError("No component error columns were available to plot.")

near_zero_adjusted_component_errors = pd.concat(near_zero_component_rows, ignore_index=True)

fig, axes = plt.subplots(2, 3, figsize=(18, 8), sharex=True)

for row_idx, group_name in enumerate(RESPONSE_GROUPS):
    group_data = near_zero_adjusted_component_errors[
        near_zero_adjusted_component_errors["Group"] == group_name
    ]

    for col_idx, component_name in enumerate(RESPONSE_GROUPS[group_name]["components"]):
        ax = axes[row_idx, col_idx]
        component_data = group_data[group_data["Component"] == component_name]

        ax.scatter(
            component_data["CaseIndex"],
            component_data["NearZeroAdjustedErrorPercent"],
            s=18,
            alpha=0.65,
        )
        ax.axhline(0.0, color="black", linewidth=0.8)
        ax.set_title(f"{group_name} {component_name}")
        ax.set_xlabel("Case index")
        ax.set_ylabel("Adjusted error (%)")
        ax.grid(True, alpha=0.3)

        zero_count = int(component_data["NearZeroResponse"].sum())
        ax.text(
            0.02,
            0.95,
            f"threshold: {COMPONENT_ZERO_THRESHOLDS[RESPONSE_GROUPS[group_name]['components'][component_name][0]]:g}\n"
            f"near-zero set to 0: {zero_count}",
            transform=ax.transAxes,
            va="top",
            fontsize=9,
        )

fig.suptitle(
    f"Component {COMPONENT_ERROR_METRIC} with component-specific zero thresholds",
    y=1.02,
)
fig.tight_layout()
plt.show()

print("Applied component-specific zero thresholds:")
for component, threshold in COMPONENT_ZERO_THRESHOLDS.items():
    zero_count = int((df[component].abs() < threshold).sum())
    print(f"  {component}: {threshold:g} ({zero_count} cases set to zero)")
print("Adjusted component error table: near_zero_adjusted_component_errors")
```

```python
# ============================================================
# PER-COMPONENT ZERO THRESHOLDS + COMBINED ERROR PLOTS
# ============================================================
# Each component has its own physical near-zero threshold.
# If |FEA response| is below that component's threshold, its
# contribution to the combined error is set to zero.
#
# Component errors are normalized by each component's 95th-
# percentile response magnitude. The group and total errors
# are RMS combinations, so strain and curvature can be combined
# without mixing their physical units directly.

from pathlib import Path

ERROR_DATA_CANDIDATES = [
    Path("AbaqusSweep/ABaqusData_with_computational_results.csv"),
    Path("../AbaqusSweep/ABaqusData_with_computational_results.csv"),
]
ERROR_DATA_PATH = next((path for path in ERROR_DATA_CANDIDATES if path.exists()), None)
if ERROR_DATA_PATH is None:
    raise FileNotFoundError("Could not find ABaqusData_with_computational_results.csv")

error_df = pd.read_csv(ERROR_DATA_PATH)

# Edit any threshold here if your physical zero tolerance changes.
COMPONENT_ZERO_THRESHOLDS = {
    "e0_11": 1e-8,   # strain
    "e0_22": 1e-8,   # strain
    "e0_12": 1e-10,  # strain; all current values are numerical-scale
    "K_11": 1e-5,    # 1/m
    "K_22": 1e-5,    # 1/m
    "K_12": 1e-5,    # 1/m
}

STRAIN_COMPONENTS = ["e0_11", "e0_22", "e0_12"]
CURVATURE_COMPONENTS = ["K_11", "K_22", "K_12"]
ALL_COMPONENTS = STRAIN_COMPONENTS + CURVATURE_COMPONENTS
COMBINED_ERROR_SCALE_PERCENTILE = 95

threshold_rows = []
adjusted_error_columns = {}

for component in ALL_COMPONENTS:
    reference = error_df[component]
    prediction = error_df[f"Comp_{component}"]
    threshold = COMPONENT_ZERO_THRESHOLDS[component]

    combined_magnitudes = np.concatenate([
        reference.abs().to_numpy(),
        prediction.abs().to_numpy(),
    ])
    scale = max(
        np.nanpercentile(combined_magnitudes, COMBINED_ERROR_SCALE_PERCENTILE),
        np.finfo(float).eps,
    )

    component_error = 100 * (prediction - reference).abs() / scale
    near_zero = reference.abs() < threshold
    adjusted_error = component_error.mask(near_zero, 0.0)

    column_name = f"AdjustedNormalizedError_{component}"
    error_df[column_name] = adjusted_error
    adjusted_error_columns[component] = column_name

    threshold_rows.append({
        "Component": component,
        "ZeroThreshold": threshold,
        "NormalizationScale": scale,
        "CasesSetToZero": int(near_zero.sum()),
        "TotalCases": len(error_df),
    })

def rms_component_error(components):
    values = error_df[[adjusted_error_columns[c] for c in components]]
    return np.sqrt((values ** 2).mean(axis=1))

error_df["StrainOnlyErrorPercent"] = rms_component_error(STRAIN_COMPONENTS)
error_df["CurvatureOnlyErrorPercent"] = rms_component_error(CURVATURE_COMPONENTS)
error_df["TotalErrorPercent"] = rms_component_error(ALL_COMPONENTS)

component_threshold_summary = pd.DataFrame(threshold_rows)
display(component_threshold_summary)

combined_error_plots = [
    ("TotalErrorPercent", "Total error", "tab:purple"),
    ("CurvatureOnlyErrorPercent", "Curvature-only error", "tab:blue"),
    ("StrainOnlyErrorPercent", "Strain-only error", "tab:orange"),
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharex=True, sharey=True)
for ax, (column, title, color) in zip(axes, combined_error_plots):
    ax.scatter(error_df.index, error_df[column], s=20, alpha=0.7, color=color)
    ax.axhline(error_df[column].median(), color="black", linestyle="--",
               linewidth=1.0, label=f"median = {error_df[column].median():.3g}%")
    ax.set_title(title)
    ax.set_xlabel("Case index")
    ax.set_ylabel("RMS normalized error (%)")
    ax.grid(True, alpha=0.3)
    ax.legend()

fig.suptitle("Combined errors after applying component-specific zero thresholds", y=1.02)
fig.tight_layout()
plt.show()

combined_error_summary = error_df[[
    "TotalErrorPercent",
    "CurvatureOnlyErrorPercent",
    "StrainOnlyErrorPercent",
]].agg(["mean", "median", "max"]).T
display(combined_error_summary)
```
